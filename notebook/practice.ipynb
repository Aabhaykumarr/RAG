```python
import pymupdf4llm
```

```python
from pathlib import Path
Path("assets").mkdir(exist_ok=True)
Path("output").mkdir(exist_ok=True)
chunks = pymupdf4llm.to_markdown(
    "RAG.pdf",
    pages = [0,1,2],
    page_chunks = True,
    write_images = True,
    image_path = "assets/",
    image_format= "png",
    dpi = 200
)
for chunk in chunks:
    page_num = chunk["metadata"] ["page_number"]
    Path(f"output/page-{page_num}.md").write_text(chunk["text"], encoding="utf8")
```
